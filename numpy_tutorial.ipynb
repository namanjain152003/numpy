```python
# execution performance
size=1_000_000

# python list
py_list=list(range(size))
start=time.time()
sq_list=[x**2 for x in py_list]
end=time.time()
print(f"time of py_list is {end-start} seconds")

# numpy array
numpy_array=np.array(py_list)
start=time.time()
sq_array=numpy_array**2
end=time.time()
print(f"time of numpy_array is {end-start} seconds")
```

```python
import sys
print(f" py_list size is {sys.getsizeof(py_list)*len(py_list)} bytes")
print(f" numpy array size is {numpy_array.nbytes} bytes")
```

```python
# operations on the numpy arrays

# RESHAPING ARRAYS
arr=np.array([[1,2,3],[4,5,6]])
print(arr , arr.shape)

reshaped=arr.reshape((3,2))
print(reshaped, reshaped.shape)

flattened=arr.flatten()  # 2D=>1D
print(flattened, flattened.shape)
```

```python
# INDEXING 
arrp= np.array([1,2,3,4,5])  #1D
print(arrp[0])
print(arrp[2])

arr=np.array([[1,2,3],[4,5,6]])
print(arr[0][2])
print(arr[1][2])

# FANCY INDEXING
arrp= np.array([1,2,3,4,5])
idx=[0,1,4]
print(arrp[idx])

# BOOLEAN MASKIng or INDEXING
print(arrp[arrp>2])
print(arrp[arrp%2==0])
```
